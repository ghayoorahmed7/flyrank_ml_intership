## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

```python
# Quick gut-check for the paragraph above: how big is the backlog, roughly?
import pandas as pd
from pathlib import Path

# Try the common places a starter repo puts this file, in order.
# Adjust CANDIDATES if your repo layout differs -- this just avoids
# hardcoding one path that only works on one machine.
CANDIDATES = [
    "data/raw/content_refresh_anonymized.csv",
    "../data/raw/content_refresh_anonymized.csv",
    "../../data/raw/content_refresh_anonymized.csv",
    "content_refresh_anonymized.csv",
]
DATA_PATH = next((p for p in CANDIDATES if Path(p).exists()), None)
if DATA_PATH is None:
    raise FileNotFoundError(
        "Couldn't find content_refresh_anonymized.csv in any of: "
        + ", ".join(CANDIDATES)
        + " -- update CANDIDATES with the real path in your repo."
    )

df = pd.read_csv(DATA_PATH)
print("Loaded from:", DATA_PATH)
print("Rows (pages):", len(df))
print("Clients:", df["client_id"].nunique())
print("Median pages per client:", int(df.groupby("client_id").size().median()))
```

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
# Back the "recoverable, human-in-the-loop" framing with a number:
# how many pages would even be in a realistic weekly review queue vs. the full backlog?
declining = df["trend_direction"] == "down"
print("Declining pages, all clients:", declining.sum())
print("If reviewing top 50/week per client, weeks to clear one client's declining backlog "
      "(median client):", round(df[declining].groupby("client_id").size().median() / 50, 1))
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# The 2-3 real numbers, computed directly from the starter CSV.

# Number 1: how common is "declining"?
trend_share = df["trend_direction"].value_counts(normalize=True).round(3)
print(trend_share)
print(f"-> {trend_share['down']*100:.1f}% of pages are trending down\n")

# Number 2: backlog size per client
declining_per_client = df[df["trend_direction"] == "down"].groupby("client_id").size()
pages_per_client = df.groupby("client_id").size()
print("Median declining pages per client:", int(declining_per_client.median()))
print("Median total pages per client:", int(pages_per_client.median()), "\n")

# Number 3: overlap of "best pages" and "currently declining"
high_impr = df["impression_tier"].isin(["good", "excellent"])
declining = df["trend_direction"] == "down"
pct = (high_impr & declining).sum() / high_impr.sum() * 100
print(f"Share of high/excellent-impression pages that are declining: {pct:.1f}%")
```
